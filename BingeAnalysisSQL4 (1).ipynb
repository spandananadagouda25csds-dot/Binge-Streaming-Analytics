```python
# Q1 - Active Revenue

query = """
SELECT
    COUNT(*) AS active_subscription_count,
    SUM(monthly_price_inr) AS total_monthly_revenue
FROM subscriptions
WHERE status = 'active'
  AND (
      end_date IS NULL
      OR end_date > '2024-06-30'
  );
"""

result = pd.read_sql(query, engine)
print(result)
```

```python
# Q2 - Signup Momentum

query = """
WITH monthly_signups AS (
    SELECT
        MONTH(signup_date) AS month,
        MONTHNAME(signup_date) AS month_name,
        COUNT(*) AS signup_count
    FROM users
    WHERE signup_date >= '2024-01-01'
      AND signup_date < '2024-07-01'
    GROUP BY MONTH(signup_date), MONTHNAME(signup_date)
),
ranked_signups AS (
    SELECT
        month,
        month_name,
        signup_count,
        ROW_NUMBER() OVER (
            ORDER BY signup_count DESC, month
        ) AS rn
    FROM monthly_signups
)
SELECT
    month,
    month_name,
    signup_count,
    MAX(
        CASE WHEN rn = 1 THEN month_name END
    ) OVER () AS highest_signup_month
FROM ranked_signups
ORDER BY month;
"""

result = pd.read_sql(query, engine)
print(result)
```

```python
# Q3 - Device Analytics

query = """
SELECT
    device_type,
    COUNT(*) AS total_sessions,
    SUM(watch_minutes) AS total_watch_minutes,
    ROUND(AVG(watch_minutes), 2) AS avg_watch_minutes,
    ROUND(
        SUM(CASE WHEN completed = 1 THEN 1 ELSE 0 END)
        * 100.0 / COUNT(*),
        2
    ) AS completion_rate
FROM watch_sessions
WHERE user_id IS NOT NULL
GROUP BY device_type
ORDER BY device_type;
"""

result = pd.read_sql(query, engine)
print(result)
```

```python
# Q4 - Rating Distribution

query = """
WITH rating_distribution AS (
    SELECT
        stars,
        COUNT(*) AS rating_count,
        ROUND(
            COUNT(*) * 100.0 / (SELECT COUNT(*) FROM ratings),
            2
        ) AS percentage
    FROM ratings
    GROUP BY stars
)
SELECT
    stars,
    rating_count,
    percentage,
    (
        SELECT ROUND(
            SUM(CASE WHEN stars IN (4, 5) THEN 1 ELSE 0 END)
            * 100.0 / COUNT(*),
            2
        )
        FROM ratings
    ) AS four_or_five_star_percentage
FROM rating_distribution
ORDER BY stars;
"""

result = pd.read_sql(query, engine)
print(result)
```

```python
# Q6 - Binge Day Detection

query = """
WITH binge_days AS (
    SELECT
        user_id,
        show_id,
        session_date,
        COUNT(*) AS session_count
    FROM watch_sessions
    WHERE user_id IS NOT NULL
      AND session_date >= '2024-04-01'
      AND session_date < '2024-07-01'
    GROUP BY user_id, show_id, session_date
    HAVING COUNT(*) >= 5
),
user_binge_counts AS (
    SELECT
        user_id,
        COUNT(*) AS binge_days
    FROM binge_days
    GROUP BY user_id
),
top_user AS (
    SELECT
        user_id,
        binge_days
    FROM user_binge_counts
    ORDER BY binge_days DESC, user_id
    LIMIT 1
)
SELECT
    (SELECT COUNT(*) FROM binge_days) AS total_binge_days,
    (SELECT user_id FROM top_user) AS top_user_id,
    (SELECT binge_days FROM top_user) AS top_user_binge_days;
"""

result = pd.read_sql(query, engine)
print(result)
```

```python
# Q7 - Q1 Signups Who Never Watched

query = """
SELECT
    COUNT(*) AS total_q1_signups,
    SUM(
        CASE
            WHEN NOT EXISTS (
                SELECT 1
                FROM watch_sessions ws
                WHERE ws.user_id = u.user_id
            )
            THEN 1
            ELSE 0
        END
    ) AS never_watched_users
FROM users u
WHERE u.signup_date >= '2024-01-01'
  AND u.signup_date < '2024-04-01';
"""

result = pd.read_sql(query, engine)
print(result)
```

```python
# Q8 - Over-Paying Premium/Family Users

query = """
WITH ranked_subscriptions AS (
    SELECT
        user_id,
        Plan,
        start_date,
        ROW_NUMBER() OVER (
            PARTITION BY user_id
            ORDER BY start_date DESC, subscription_id DESC
        ) AS rn
    FROM subscriptions
    WHERE status = 'active'
      AND start_date <= '2024-06-30'
      AND (
          end_date IS NULL
          OR end_date > '2024-06-30'
      )
)
SELECT
    COUNT(*) AS overpaying_users
FROM ranked_subscriptions rs
WHERE rs.rn = 1
  AND rs.Plan IN ('Premium', 'Family')
  AND EXISTS (
      SELECT 1
      FROM watch_sessions ws
      WHERE ws.user_id = rs.user_id
  )
  AND NOT EXISTS (
      SELECT 1
      FROM watch_sessions ws
      JOIN shows s
        ON ws.show_id = s.show_id
      WHERE ws.user_id = rs.user_id
        AND s.min_plan IN ('Premium', 'Family')
  );
"""

result = pd.read_sql(query, engine)
print(result)
```

```python
# Q9 - Upgrade Success Cohort

query = """
WITH ranked_subscriptions AS (
    SELECT
        user_id,
        Plan,
        start_date,
        ROW_NUMBER() OVER (
            PARTITION BY user_id
            ORDER BY start_date, subscription_id
        ) AS rn
    FROM subscriptions
),
first_basic AS (
    SELECT
        user_id,
        start_date AS basic_start_date
    FROM ranked_subscriptions
    WHERE rn = 1
      AND Plan = 'Basic'
),
first_upgrade AS (
    SELECT
        fb.user_id,
        MIN(s.start_date) AS first_upgrade_date
    FROM first_basic fb
    JOIN subscriptions s
      ON s.user_id = fb.user_id
     AND s.start_date > fb.basic_start_date
     AND s.Plan IN ('Premium', 'Family')
    GROUP BY fb.user_id
),
eligible_users AS (
    SELECT
        u.user_id,
        DATEDIFF(fu.first_upgrade_date, u.signup_date)
            AS days_to_upgrade
    FROM users u
    JOIN first_upgrade fu
      ON u.user_id = fu.user_id
    WHERE u.signup_date >= '2024-01-01'
      AND u.signup_date < '2024-02-01'
      AND EXISTS (
          SELECT 1
          FROM subscriptions s
          WHERE s.user_id = u.user_id
            AND s.status = 'active'
            AND s.start_date <= '2024-06-30'
            AND (
                s.end_date IS NULL
                OR s.end_date > '2024-06-30'
            )
      )
)
SELECT
    COUNT(*) AS total_upgraded_users,
    ROUND(AVG(days_to_upgrade), 2) AS avg_days_to_upgrade
FROM eligible_users;
"""

result = pd.read_sql(query, engine)
print(result)
```

## Q9 – Upgrade Success Cohort

- Average days from signup to first upgrade: **64.96 days**

A total of 55 users who signed up in January 2024 started with a Basic plan and later upgraded to Premium or Family while remaining active as of 30 June 2024.

```python
# Q10 - Cliffhanger Comebacks

query = """
WITH comeback_events AS (
    SELECT DISTINCT
        first_session.user_id,
        first_session.show_id,
        first_session.session_date AS incomplete_date
    FROM watch_sessions first_session
    JOIN watch_sessions second_session
      ON first_session.user_id = second_session.user_id
     AND first_session.show_id = second_session.show_id
     AND second_session.session_date BETWEEN
         DATE_ADD(first_session.session_date, INTERVAL 1 DAY)
         AND DATE_ADD(first_session.session_date, INTERVAL 7 DAY)
    WHERE first_session.user_id IS NOT NULL
      AND first_session.completed = 0
),
show_comebacks AS (
    SELECT
        ce.show_id,
        s.title,
        COUNT(*) AS comeback_count
    FROM comeback_events ce
    JOIN shows s
      ON ce.show_id = s.show_id
    GROUP BY ce.show_id, s.title
),
top_show AS (
    SELECT
        show_id,
        title,
        comeback_count
    FROM show_comebacks
    ORDER BY comeback_count DESC, show_id
    LIMIT 1
)
SELECT
    (SELECT COUNT(*) FROM comeback_events)
        AS total_comeback_events,
    (SELECT show_id FROM top_show)
        AS top_show_id,
    (SELECT title FROM top_show)
        AS top_show_title,
    (SELECT comeback_count FROM top_show)
        AS top_show_comeback_count;
"""

result = pd.read_sql(query, engine)
print(result)
```

```python
# Q11 - Consecutive-Week Engagement

query = """
WITH weekly_activity AS (
    SELECT DISTINCT
        user_id,
        DATE_SUB(
            session_date,
            INTERVAL WEEKDAY(session_date) DAY
        ) AS week_start
    FROM watch_sessions
    WHERE user_id IS NOT NULL
),
numbered_weeks AS (
    SELECT
        user_id,
        week_start,
        ROW_NUMBER() OVER (
            PARTITION BY user_id
            ORDER BY week_start
        ) AS rn
    FROM weekly_activity
),
streak_groups AS (
    SELECT
        user_id,
        (
            TO_DAYS(week_start) DIV 7
        ) - rn AS streak_group
    FROM numbered_weeks
),
streaks AS (
    SELECT
        user_id,
        COUNT(*) AS streak_weeks
    FROM streak_groups
    GROUP BY user_id, streak_group
),
user_streaks AS (
    SELECT
        user_id,
        MAX(streak_weeks) AS longest_user_streak
    FROM streaks
    GROUP BY user_id
)
SELECT
    SUM(
        CASE
            WHEN longest_user_streak >= 4 THEN 1
            ELSE 0
        END
    ) AS users_with_4_week_streak,
    MAX(longest_user_streak) AS longest_streak_weeks,
    (
        SELECT user_id
        FROM user_streaks
        ORDER BY longest_user_streak DESC, user_id
        LIMIT 1
    ) AS one_user_with_longest_streak
FROM user_streaks;
"""

result = pd.read_sql(query, engine)
print(result)
```

- Longest streak: **26 weeks**

A total of 1,675 users watched at least one session in four or more consecutive calendar weeks. The longest recorded engagement streak was 26 weeks, achieved by user U00211.

```python
# Q12 - Churn Signal Detection

query = """
WITH monthly_watch AS (
    SELECT
        user_id,
        SUM(
            CASE
                WHEN session_date >= '2024-05-01'
                 AND session_date < '2024-06-01'
                THEN watch_minutes
                ELSE 0
            END
        ) AS may_minutes,
        SUM(
            CASE
                WHEN session_date >= '2024-06-01'
                 AND session_date < '2024-07-01'
                THEN watch_minutes
                ELSE 0
            END
        ) AS june_minutes
    FROM watch_sessions
    WHERE user_id IS NOT NULL
      AND session_date >= '2024-05-01'
      AND session_date < '2024-07-01'
    GROUP BY user_id
),
churn_users AS (
    SELECT
        mw.user_id,
        u.Name,
        mw.may_minutes,
        mw.june_minutes,
        ROUND(
            (mw.may_minutes - mw.june_minutes)
            * 100.0 / mw.may_minutes,
            2
        ) AS drop_percentage
    FROM monthly_watch mw
    JOIN users u
      ON mw.user_id = u.user_id
    WHERE mw.may_minutes > 0
      AND mw.june_minutes <= mw.may_minutes * 0.5
)
SELECT
    user_id,
    Name,
    may_minutes,
    june_minutes,
    drop_percentage,
    COUNT(*) OVER () AS total_churn_signal_users
FROM churn_users
ORDER BY drop_percentage DESC, user_id;
"""

result = pd.read_sql(query, engine)
print(result)
```

## Q12 – Churn Signal Detection

- Total churn signal users: **521**

A total of 521 users experienced a 50% or greater reduction in watch time in June 2024 compared with May 2024. These users are potential targets for retention campaigns.
